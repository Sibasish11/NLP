## 1. Regular Expressions

```python
import re

text = "Contact us at hello@example.com or visit https://example.com. Call +91-9876543210."

emails = re.findall(r"[\w.+-]+@[\w-]+\.[\w.-]+", text)
urls = re.findall(r"https?://[\w./-]+", text)
phones = re.findall(r"\+?\d[\d-]{8,}\d", text)

print("Emails:", emails)
print("URLs:", urls)
print("Phones:", phones)
```

## 2. Tokenization

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize, sent_tokenize

sample = "Dr. Smith went to Washington D.C. He didn't stay long, though."

print("Sentences:", sent_tokenize(sample))
print("Words:", word_tokenize(sample))
```

## 3. Stemming vs. Lemmatization

```python
nltk.download('wordnet')

from nltk.stem import PorterStemmer, WordNetLemmatizer

words = ["studies", "studying", "cries", "cried", "better", "running"]

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

for w in words:
    print(f"{w:12s} -> stem: {stemmer.stem(w):10s} | lemma: {lemmatizer.lemmatize(w, pos='v')}")
```

## 4. Edit Distance (Levenshtein) — implemented from scratch

```python
def edit_distance(s1: str, s2: str) -> int:
    m, n = len(s1), len(s2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            cost = 0 if s1[i - 1] == s2[j - 1] else 1
            dp[i][j] = min(
                dp[i - 1][j] + 1,       # deletion
                dp[i][j - 1] + 1,       # insertion
                dp[i - 1][j - 1] + cost # substitution
            )
    return dp[m][n]


print(edit_distance("kitten", "sitting"))  # expect 3
print(edit_distance("intention", "execution"))  # expect 5
```
